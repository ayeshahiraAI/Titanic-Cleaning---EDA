```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data=pd.read_csv('/content/train.csv')
```

```python
data = data.drop_duplicates()
```

```python

data['Age'].fillna(data['Age'].mean(), inplace=True)
data['Embarked'].fillna(data['Embarked'].mode()[0], inplace=True)



```

```python
data.drop(columns=['Cabin'], inplace=True)
```

```python
data.columns = data.columns.str.lower()
```

```python
data.describe()
```

```python
num_cols = data.select_dtypes(include=["int64", "float64"]).columns
for col in num_cols:
    data[col].fillna(data[col].median(), inplace=True)

```

```python
# Categorical columns → mode
cat_cols = data.select_dtypes(include=["object"]).columns
for col in cat_cols:
    data[col].fillna(data[col].mode()[0], inplace=True)
```

```python
for col in cat_cols:
    data[col] = data[col].str.strip()
```

```python
if "Date" in data.columns:
    data["Date"] = pd.to_datetime(data["Date"], errors="coerce")
```

```python
print(data.isnull().sum())
print(data.head())
```

```python
# Save cleaned dataset
data.to_csv("/content/train.csv", index=False)

print("Data cleaning completed. Cleaned file saved.")
```

```python
sns.set(style="whitegrid")
```

```python
# Basic Statistics
print("Mean:\n", data.mean(numeric_only=True))
print("Median:\n", data.median(numeric_only=True))
print("Minimum:\n", data.min(numeric_only=True))
print("Maximum:\n", data.max(numeric_only=True))
```

```python
#  Correlation Analysis
plt.figure(figsize=(10,6))
sns.heatmap(data.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
#  Outlier Detection
for col in num_cols:
    plt.figure(figsize=(6,4))
    sns.boxplot(x=data[col])
    plt.title(f"Outliers in {col}")
    plt.show()
```

```python
# Data Visualizations
# Histogram
data[num_cols].hist(figsize=(10,8))
plt.suptitle("Histograms of Numerical Features")
plt.show()

```

```python
# Line Graph (time-based data)
if "Date" in data.columns and len(num_cols) > 0:
    data.groupby("Date")[num_cols[0]].mean().plot()
    plt.title("Line Graph Showing Trend Over Time")
    plt.show()
```

```python
cat_cols = ['passengerid', 'survived', 'pclass', 'age', 'sibsp',
       'parch','fare']

# -----------------------------
# Plot bar charts
# -----------------------------
sns.set(style="whitegrid")

for col in cat_cols:
    plt.figure(figsize=(6,4))
    sns.countplot(x=col, data=data, color='skyblue')

    plt.title(f"Bar Chart of {col}")
    plt.ylabel("Count")
    plt.show()
```

```python
sns.countplot(x='sex', hue='survived', data=data, palette='pastel')
plt.title('Count of Sex by Survival')
plt.show()
```
